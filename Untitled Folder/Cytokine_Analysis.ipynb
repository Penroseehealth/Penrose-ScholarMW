```python
# Select the early- and mid-pregnancy Firmicutes measurements
time_columns = [
    "Firmicutes_Early_Pregnancy_%",
    "Firmicutes_Mid_Pregnancy_%"
]

# Reshape the data so each measurement has its own row
firmicutes_long = df[
    ["Malaria_Group"] + time_columns
].melt(
    id_vars="Malaria_Group",
    var_name="Timepoint",
    value_name="Firmicutes_Percent"
)

# Give the timepoints shorter, readable labels
firmicutes_long["Timepoint"] = firmicutes_long["Timepoint"].map({
    "Firmicutes_Early_Pregnancy_%": "Early pregnancy",
    "Firmicutes_Mid_Pregnancy_%": "Mid-pregnancy"
})

# Ensure measurements are numeric
firmicutes_long["Firmicutes_Percent"] = pd.to_numeric(
    firmicutes_long["Firmicutes_Percent"],
    errors="coerce"
)

# Calculate mean Firmicutes percentage for each group and timepoint
firmicutes_summary = (
    firmicutes_long
    .groupby(["Malaria_Group", "Timepoint"], observed=True)
    ["Firmicutes_Percent"]
    .mean()
    .reset_index(name="Mean_Firmicutes")
)

# Ensure early pregnancy appears before mid-pregnancy
time_order = ["Early pregnancy", "Mid-pregnancy"]
firmicutes_summary["Timepoint"] = pd.Categorical(
    firmicutes_summary["Timepoint"],
    categories=time_order,
    ordered=True
)
firmicutes_summary = firmicutes_summary.sort_values("Timepoint")

# Draw the line graph
plt.figure(figsize=(8, 5))

for group, group_data in firmicutes_summary.groupby(
    "Malaria_Group", observed=True
):
    plt.plot(
        group_data["Timepoint"],
        group_data["Mean_Firmicutes"],
        marker="o",
        linewidth=2,
        label=str(group)
    )

plt.title("Mean Firmicutes Percentage by Malaria Group")
plt.xlabel("Pregnancy timepoint")
plt.ylabel("Mean Firmicutes (%)")
plt.legend(title="Malaria group")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import t

# Prepare the required variables
plot_df = df[
    ["Malaria_Group", "Birth_Weight_kg"]
].copy()

plot_df["Birth_Weight_kg"] = pd.to_numeric(
    plot_df["Birth_Weight_kg"],
    errors="coerce"
)

plot_df = plot_df.dropna(
    subset=["Malaria_Group", "Birth_Weight_kg"]
)

# Calculate the mean, sample size and standard deviation
summary = (
    plot_df.groupby("Malaria_Group", observed=True)["Birth_Weight_kg"]
    .agg(["count", "mean", "std"])
    .reset_index()
)

# Calculate 95% confidence intervals for each group mean
summary["SE"] = summary["std"] / np.sqrt(summary["count"])

summary["CI95"] = (
    t.ppf(0.975, summary["count"] - 1) * summary["SE"]
)

display(summary.round(3))

# Draw the bar graph
sns.set_theme(style="whitegrid")

fig, ax = plt.subplots(figsize=(9, 6), dpi=180)

bars = ax.bar(
    summary["Malaria_Group"].astype(str),
    summary["mean"],
    yerr=summary["CI95"],
    capsize=7,
    color=sns.color_palette(
        "colorblind", n_colors=len(summary)
    ),
    edgecolor="black",
    linewidth=0.8
)

# Add the mean value above each bar
for bar, mean in zip(bars, summary["mean"]):
    ax.annotate(
        f"{mean:.2f} kg",
        (bar.get_x() + bar.get_width() / 2, bar.get_height()),
        xytext=(0, 9),
        textcoords="offset points",
        ha="center",
        fontsize=10,
        fontweight="bold"
    )

ax.set_title(
    "Mean Birth Weight by Malaria Group",
    fontsize=15,
    fontweight="bold",
    pad=15
)

ax.set_xlabel("Malaria group", fontsize=12)
ax.set_ylabel("Mean birth weight (kg)", fontsize=12)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()

# Save a high-resolution image and PDF
fig.savefig(
    "mean_birth_weight_by_malaria.png",
    dpi=300,
    bbox_inches="tight"
)

fig.savefig(
    "mean_birth_weight_by_malaria.pdf",
    bbox_inches="tight"
)

plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import textwrap

from matplotlib.colors import Normalize
from matplotlib.cm import ScalarMappable
from matplotlib.lines import Line2D
from matplotlib.patches import Patch
from scipy.stats import chi2_contingency

# ==========================================================
# 1. CONFIGURATION
# ==========================================================

central_col = "Malaria_Group"

if central_col not in df.columns:
    raise KeyError(
        f"{central_col} was not found. Check df.columns.tolist()."
    )

if df[central_col].nunique(dropna=True) < 2:
    raise ValueError(
        "Placental malaria must have at least two observed groups."
    )

# Exclude identifiers because they do not represent biological
# or clinical measurements.
id_cols = [
    col for col in df.columns
    if col.lower() in {
        "id", "participant_id", "patient_id",
        "subject_id", "record_id"
    }
    or col.lower().endswith("_id")
]

variables = [
    col for col in df.columns
    if col != central_col and col not in id_cols
]

print("Central variable:", central_col)
print("Malaria groups:", df[central_col].dropna().unique().tolist())
print("Research variables to map:", len(variables))


# ==========================================================
# 2. GROUP VARIABLES INTO THEMATIC DOMAINS
# ==========================================================

def identify_theme(variable):
    name = variable.lower()

    if any(x in name for x in [
        "firmicutes", "bacteroidetes", "microbiome",
        "microbiota", "gut", "bacterial", "bacteria"
    ]):
        return "Microbiome"

    if any(x in name for x in [
        "cytokine", "interleukin", "tnf", "ifn",
        "chemokine", "il_", "il-"
    ]):
        return "Cytokines"

    if any(x in name for x in [
        "hormone", "progesterone", "estrogen",
        "estradiol", "hcg", "lactogen"
    ]):
        return "Placental hormones"

    if any(x in name for x in [
        "hemozoin", "parasite", "placenta", "malaria"
    ]):
        return "Placental/parasite markers"

    if any(x in name for x in [
        "birth", "gestational", "delivery",
        "neonatal", "iugr", "growth", "weight"
    ]):
        return "Birth outcomes"

    if any(x in name for x in [
        "maternal", "mother", "parity", "gravidity",
        "age", "bmi"
    ]):
        return "Maternal characteristics"

    return "Other measures"


theme_colours = {
    "Microbiome": "#168C83",
    "Cytokines": "#D95F59",
    "Placental hormones": "#8867B5",
    "Placental/parasite markers": "#4575A5",
    "Birth outcomes": "#E69F00",
    "Maternal characteristics": "#6096BA",
    "Other measures": "#8993A4"
}

theme_order = list(theme_colours.keys())


# ==========================================================
# 3. CALCULATE ASSOCIATION STRENGTH WITH MALARIA STATUS
# ==========================================================

results = []

for col in variables:

    pair = df[[central_col, col]].dropna().copy()
    n = len(pair)

    strength = np.nan
    measure = "Not estimable"

    if (
        n >= 3
        and pair[central_col].nunique() >= 2
        and pair[col].nunique() >= 2
    ):

        # Recognise numeric measurements, including numeric strings
        # when most values can be converted and there are many
        # distinct measurements.
        numeric_values = pd.to_numeric(
            pair[col], errors="coerce"
        )

        numeric_variable = (
            pd.api.types.is_numeric_dtype(df[col])
            or (
                numeric_values.notna().mean() >= 0.90
                and numeric_values.nunique() >= 8
            )
        )

        if numeric_variable:

            numeric_pair = pd.DataFrame({
                "group": pair[central_col],
                "value": numeric_values
            }).dropna()

            grand_mean = numeric_pair["value"].mean()

            total_variation = (
                (numeric_pair["value"] - grand_mean) ** 2
            ).sum()

            between_group_variation = sum(
                len(group_data)
                * (group_data["value"].mean() - grand_mean) ** 2
                for _, group_data in numeric_pair.groupby(
                    "group", observed=True
                )
            )

            if total_variation > 0:
                strength = np.sqrt(
                    max(
                        0,
                        between_group_variation / total_variation
                    )
                )
                measure = "Eta (η)"

        else:

            contingency = pd.crosstab(
                pair[central_col],
                pair[col]
            )

            if (
                contingency.shape[0] >= 2
                and contingency.shape[1] >= 2
            ):
                chi2 = chi2_contingency(
                    contingency,
                    correction=False
                )[0]

                total_n = contingency.to_numpy().sum()

                denominator = total_n * min(
                    contingency.shape[0] - 1,
                    contingency.shape[1] - 1
                )

                if denominator > 0:
                    strength = np.sqrt(chi2 / denominator)
                    measure = "Cramer's V"

    results.append({
        "Variable": col,
        "Theme": identify_theme(col),
        "Measure": measure,
        "Strength": strength,
        "N": n
    })


associations = pd.DataFrame(results)

associations["ThemeOrder"] = associations["Theme"].map(
    {theme: i for i, theme in enumerate(theme_order)}
)

# Arrange related variables near each other around the network.
associations = associations.sort_values(
    ["ThemeOrder", "Strength"],
    ascending=[True, False],
    na_position="last"
).reset_index(drop=True)

print("\nAssociation results:")
display(
    associations[
        ["Variable", "Theme", "Measure", "Strength", "N"]
    ].style.format({"Strength": "{:.3f}"}, na_rep="Not estimable")
)


# ==========================================================
# 4. DRAW THE NETWORK
# ==========================================================

fig, ax = plt.subplots(figsize=(20, 20), dpi=160)

radius = 7.2
label_radius = 8.0

edge_cmap = plt.cm.viridis
edge_norm = Normalize(vmin=0, vmax=1)

# Position the research variables in a circle.
number_of_variables = len(associations)

if number_of_variables == 0:
    raise ValueError("No research variables are available to plot.")

positions = {}

for i, row in associations.iterrows():

    angle = np.pi / 2 - (
        2 * np.pi * i / number_of_variables
    )

    x = radius * np.cos(angle)
    y = radius * np.sin(angle)

    positions[row["Variable"]] = (x, y)

    # Draw each connection from the central malaria node.
    if pd.notna(row["Strength"]):

        edge_colour = edge_cmap(
            edge_norm(row["Strength"])
        )

        line_width = 1.0 + 3.5 * row["Strength"]
        line_style = "-"
        alpha = 0.85

    else:

        edge_colour = "#B8B8B8"
        line_width = 1.0
        line_style = "--"
        alpha = 0.65

    ax.plot(
        [0, x],
        [0, y],
        color=edge_colour,
        linewidth=line_width,
        linestyle=line_style,
        alpha=alpha,
        zorder=1
    )

# Draw the surrounding variable nodes and their labels.
for _, row in associations.iterrows():

    col = row["Variable"]
    theme = row["Theme"]
    x, y = positions[col]

    ax.scatter(
        x, y,
        s=390,
        color=theme_colours[theme],
        edgecolor="white",
        linewidth=1.8,
        zorder=3
    )

    # Make variable names easier to read.
    label = col.replace("_", " ").replace("%", "(%)")
    label = textwrap.fill(label, width=22)

    # Add the association coefficient beside the variable name.
    if pd.isna(row["Strength"]):
        score_text = "Association: N/A"
    elif row["Measure"] == "Eta (η)":
        score_text = f"η = {row['Strength']:.2f}"
    else:
        score_text = f"V = {row['Strength']:.2f}"

    label_text = f"{label}\n{score_text}"

    lx = label_radius * np.cos(
        np.arctan2(y, x)
    )
    ly = label_radius * np.sin(
        np.arctan2(y, x)
    )

    if x > 1:
        horizontal_alignment = "left"
    elif x < -1:
        horizontal_alignment = "right"
    else:
        horizontal_alignment = "center"

    ax.text(
        lx, ly,
        label_text,
        ha=horizontal_alignment,
        va="center",
        fontsize=8.5,
        linespacing=1.2,
        color="#202938",
        zorder=4
    )

# Draw the central placental malaria node.
ax.scatter(
    0, 0,
    s=3200,
    color="#17365D",
    edgecolor="white",
    linewidth=3,
    zorder=5
)

ax.text(
    0, 0,
    "PLACENTAL\nMALARIA\nSTATUS",
    ha="center",
    va="center",
    fontsize=14,
    fontweight="bold",
    color="white",
    zorder=6
)

# Main title
fig.suptitle(
    "Research Variable Relationships Centred on Placental Malaria",
    fontsize=19,
    fontweight="bold",
    y=0.97
)

# Explain connection strength with a colour scale.
sm = ScalarMappable(norm=edge_norm, cmap=edge_cmap)
sm.set_array([])

cbar = fig.colorbar(
    sm,
    ax=ax,
    shrink=0.48,
    pad=0.025
)

cbar.set_label(
    "Association strength (0 = weaker, 1 = stronger)",
    fontsize=10
)

# Add a legend for variable domains.
legend_handles = [
    Patch(
        facecolor=theme_colours[theme],
        label=theme
    )
    for theme in theme_order
    if theme in associations["Theme"].values
]

fig.legend(
    handles=legend_handles,
    loc="lower center",
    ncol=3,
    frameon=False,
    fontsize=10,
    bbox_to_anchor=(0.5, 0.015)
)

ax.set_xlim(-10.7, 10.7)
ax.set_ylim(-10.7, 10.7)
ax.set_aspect("equal")
ax.axis("off")

plt.subplots_adjust(top=0.92, bottom=0.07)

# Save a sharp image and a scalable PDF.
fig.savefig(
    "placental_malaria_relationship_network.png",
    dpi=300,
    bbox_inches="tight"
)

fig.savefig(
    "placental_malaria_relationship_network.pdf",
    bbox_inches="tight"
)

# Export the association table as well.
associations.drop(
    columns=["ThemeOrder"]
).to_excel(
    "placental_malaria_associations.xlsx",
    index=False
)

plt.show()

print("\nFiles saved:")
print("1. placental_malaria_relationship_network.png")
print("2. placental_malaria_relationship_network.pdf")
print("3. placental_malaria_associations.xlsx")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from matplotlib.patches import Circle, Ellipse, FancyArrowPatch, Patch
from matplotlib.lines import Line2D
import textwrap

# =========================================================
# 1. PREPARE THE DATA
# =========================================================

CENTRAL = "Malaria_Group"

if "df" not in globals():
    raise NameError("Load your Excel dataset into a DataFrame named df first.")

if CENTRAL not in df.columns:
    raise KeyError(
        f"Column '{CENTRAL}' was not found. Run print(df.columns.tolist())."
    )

# Exclude records with no malaria-group classification.
data = df.loc[df[CENTRAL].notna()].copy()
data[CENTRAL] = data[CENTRAL].astype(str)

groups = sorted(data[CENTRAL].unique())

if len(groups) < 2:
    raise ValueError("At least two malaria groups are required.")

# Exclude identifiers from biological measurements.
id_columns = [
    c for c in data.columns
    if c.lower().endswith("_id")
    or c.lower() in {"id", "participant", "participantid"}
]

# Discover variables with numeric measurements.
def find_numeric_columns(keywords):
    found = []

    for column in data.columns:
        if column == CENTRAL or column in id_columns:
            continue

        name = column.lower()

        if not any(word in name for word in keywords):
            continue

        values = pd.to_numeric(data[column], errors="coerce")

        if values.notna().sum() >= 3 and values.nunique() > 1:
            found.append(column)

    return found


firmicutes_cols = find_numeric_columns(["firmicutes"])

cytokine_cols = find_numeric_columns([
    "cytokine", "tnf", "interleukin", "interferon",
    "ifn", "il-", "il_", "il6", "il10", "chemokine"
])

hormone_cols = find_numeric_columns([
    "hormone", "progesterone", "estrogen", "oestrogen",
    "estradiol", "hcg", "lactogen", "hpl"
])

placenta_cols = find_numeric_columns([
    "hemozoin", "placenta", "placental",
    "parasite", "infected", "infection"
])

# Prevent markers from appearing in more than one domain.
placenta_cols = [
    c for c in placenta_cols
    if c not in firmicutes_cols
    and c not in cytokine_cols
    and c not in hormone_cols
]

domains = {
    "Firmicutes": firmicutes_cols,
    "Placental infection markers": placenta_cols,
    "Cytokines": cytokine_cols,
    "Placental hormones": hormone_cols
}

print("Malaria groups:", groups)

for domain, columns in domains.items():
    print(f"\n{domain}:")
    print(columns if columns else "No matching numeric columns detected.")


# =========================================================
# 2. GRAPH STYLE
# =========================================================

group_palette = plt.get_cmap("tab10")
group_colours = {
    group: group_palette(i % 10)
    for i, group in enumerate(groups)
}

domain_colours = {
    "Firmicutes": "#138D83",
    "Placental infection markers": "#D95770",
    "Cytokines": "#8964B5",
    "Placental hormones": "#E39A16"
}

plt.rcParams["figure.facecolor"] = "white"
plt.rcParams["axes.facecolor"] = "white"
plt.rcParams["font.family"] = "DejaVu Sans"

fig = plt.figure(figsize=(19, 12), dpi=160)

layout = fig.add_gridspec(
    2, 3,
    width_ratios=[1.4, 1.0, 1.4],
    height_ratios=[1, 1],
    wspace=0.28,
    hspace=0.38
)

ax_firm = fig.add_subplot(layout[0, 0])
ax_cyto = fig.add_subplot(layout[1, 0])
ax_center = fig.add_subplot(layout[:, 1])
ax_placenta = fig.add_subplot(layout[0, 2])
ax_hormone = fig.add_subplot(layout[1, 2])


# =========================================================
# 3. HELPER FUNCTIONS FOR THE DATA CHARTS
# =========================================================

def nice_label(column):
    return textwrap.fill(
        column.replace("_", " ").replace("%", "(%)"),
        width=18
    )


def style_panel(ax, title, colour):
    ax.set_title(
        title,
        fontsize=13,
        fontweight="bold",
        color=colour,
        pad=12
    )
    ax.grid(axis="y", alpha=0.22)
    ax.set_axisbelow(True)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)


def plot_firmicutes(ax, columns):
    style_panel(
        ax,
        "GUT MICROBIOME — FIRMICUTES",
        domain_colours["Firmicutes"]
    )

    if not columns:
        ax.text(
            0.5, 0.5,
            "No numeric Firmicutes measurements found.\n"
            "Check the printed column names.",
            ha="center", va="center",
            transform=ax.transAxes
        )
        ax.set_axis_off()
        return

    # Prefer the biological order of repeated measurements.
    def time_order(column):
        name = column.lower()
        for index, word in enumerate(["early", "mid", "middle", "late"]):
            if word in name:
                return index
        return 10

    columns = sorted(columns, key=time_order)

    x = np.arange(len(columns))

    # If early/mid measurements exist, show how group means change.
    if len(columns) >= 2:
        for group in groups:
            subset = data[data[CENTRAL] == group]

            means = []
            errors = []

            for column in columns:
                values = pd.to_numeric(
                    subset[column], errors="coerce"
                ).dropna()

                means.append(values.mean() if len(values) else np.nan)

                if len(values) > 1:
                    errors.append(
                        values.std(ddof=1) / np.sqrt(len(values))
                    )
                else:
                    errors.append(0)

            ax.errorbar(
                x, means, yerr=errors,
                marker="o", markersize=7,
                linewidth=2.4, capsize=4,
                color=group_colours[group],
                label=group
            )

        ax.set_xticks(x)
        ax.set_xticklabels(
            [nice_label(c) for c in columns],
            fontsize=8
        )
        ax.set_ylabel("Mean measured abundance (%)")
        ax.legend(title="Malaria group", fontsize=8)

    else:
        # With only one measurement, use a grouped mean bar chart.
        column = columns[0]
        width = 0.75 / len(groups)

        for i, group in enumerate(groups):
            values = pd.to_numeric(
                data.loc[data[CENTRAL] == group, column],
                errors="coerce"
            ).dropna()

            mean = values.mean() if len(values) else np.nan
            sem = (
                values.std(ddof=1) / np.sqrt(len(values))
                if len(values) > 1 else 0
            )

            ax.bar(
                i, mean, width=0.65,
                color=group_colours[group],
                yerr=sem, capsize=4
            )

        ax.set_xticks([0])
        ax.set_xticklabels([nice_label(column)])
        ax.set_ylabel("Mean measured abundance (%)")

    ax.set_xlabel("Measurement / pregnancy timepoint")


def plot_standardized_markers(ax, columns, title, colour):
    style_panel(ax, title, colour)

    if not columns:
        ax.text(
            0.5, 0.5,
            "No matching numeric measurements found.\n"
            "Review the column names printed above.",
            ha="center", va="center",
            transform=ax.transAxes
        )
        ax.set_axis_off()
        return

    # Standardise each marker separately because biomarkers can
    # have different measurement units and ranges.
    results = []

    for column in columns:
        values = pd.to_numeric(data[column], errors="coerce")
        sd = values.std(ddof=0)

        if not np.isfinite(sd) or sd == 0:
            continue

        z = (values - values.mean()) / sd

        for group in groups:
            group_z = z[data[CENTRAL] == group].dropna()

            results.append({
                "Variable": column,
                "Group": group,
                "Mean_Z": group_z.mean() if len(group_z) else np.nan
            })

    summary = pd.DataFrame(results)

    if summary.empty:
        ax.text(
            0.5, 0.5,
            "No markers with usable variation.",
            ha="center", va="center",
            transform=ax.transAxes
        )
        ax.set_axis_off()
        return

    usable = [
        column for column in columns
        if column in summary["Variable"].values
    ]

    x = np.arange(len(usable))
    width = min(0.8 / len(groups), 0.35)

    for i, group in enumerate(groups):
        group_summary = (
            summary[summary["Group"] == group]
            .set_index("Variable")["Mean_Z"]
            .reindex(usable)
        )

        offset = (i - (len(groups) - 1) / 2) * width

        ax.bar(
            x + offset,
            group_summary.values,
            width=width,
            color=group_colours[group],
            label=group
        )

    ax.axhline(0, color="#333333", linewidth=1)
    ax.set_xticks(x)
    ax.set_xticklabels(
        [nice_label(c) for c in usable],
        rotation=35,
        ha="right",
        fontsize=8
    )
    ax.set_ylabel("Mean standardised value (z-score)")
    ax.legend(title="Malaria group", fontsize=8)


def plot_placental_panel(ax, columns):
    style_panel(
        ax,
        "INFECTED PLACENTA / PARASITE MARKERS",
        domain_colours["Placental infection markers"]
    )

    if columns:
        plot_standardized_markers(
            ax,
            columns,
            "INFECTED PLACENTA / PARASITE MARKERS",
            domain_colours["Placental infection markers"]
        )
        return

    # If no separate numeric placental marker is available,
    # show the actual participant counts in the malaria groups.
    counts = data[CENTRAL].value_counts().reindex(groups, fill_value=0)

    bars = ax.bar(
        groups,
        counts.values,
        color=[group_colours[g] for g in groups],
        edgecolor="white"
    )

    ax.bar_label(bars, padding=3, fontsize=10)
    ax.set_ylabel("Participants (n)")
    ax.set_xlabel("Recorded malaria group")

    ax.text(
        0.5, -0.27,
        "No separate numeric hemozoin or placental marker was detected.\n"
        "These bars show group sizes, not histological findings.",
        transform=ax.transAxes,
        ha="center",
        va="top",
        fontsize=8,
        color="#555555",
        wrap=True
    )


# =========================================================
# 4. DRAW THE FOUR BIOLOGICAL DATA PANELS
# =========================================================

plot_firmicutes(ax_firm, firmicutes_cols)

plot_standardized_markers(
    ax_cyto,
    cytokine_cols,
    "CYTOKINES",
    domain_colours["Cytokines"]
)

plot_placental_panel(ax_placenta, placenta_cols)

plot_standardized_markers(
    ax_hormone,
    hormone_cols,
    "PLACENTAL HORMONES",
    domain_colours["Placental hormones"]
)


# =========================================================
# 5. CREATE THE CENTRAL PLACENTAL MALARIA ILLUSTRATION
# =========================================================

ax_center.set_xlim(0, 1)
ax_center.set_ylim(0, 1)
ax_center.axis("off")

ax_center.text(
    0.5, 0.96,
    "BIOLOGICAL RELATIONSHIP MAP",
    ha="center", va="top",
    fontsize=13, fontweight="bold",
    color="#17365D"
)

# Draw a circular background around the central placenta.
ax_center.add_patch(
    Circle(
        (0.5, 0.53), 0.31,
        facecolor="#FFF0EE",
        edgecolor="#D95770",
        linewidth=2.5,
        zorder=1
    )
)

# Stylised placenta: illustrative vector shapes, not a photograph.
ax_center.add_patch(
    Ellipse(
        (0.5, 0.61),
        width=0.31,
        height=0.25,
        angle=-12,
        facecolor="#E98A92",
        edgecolor="#A8314D",
        linewidth=2,
        zorder=2
    )
)

ax_center.add_patch(
    Ellipse(
        (0.5, 0.61),
        width=0.24,
        height=0.18,
        angle=-12,
        facecolor="#F6B1A5",
        edgecolor="#C34C62",
        linewidth=1.4,
        zorder=3
    )
)

# Stylised branching vessels.
vessel_segments = [
    [(0.50, 0.61), (0.47, 0.65), (0.43, 0.67)],
    [(0.49, 0.61), (0.45, 0.60), (0.42, 0.57)],
    [(0.50, 0.61), (0.54, 0.65), (0.57, 0.66)],
    [(0.51, 0.61), (0.55, 0.60), (0.58, 0.57)],
    [(0.50, 0.62), (0.50, 0.57), (0.48, 0.54)],
    [(0.51, 0.61), (0.52, 0.57), (0.55, 0.55)]
]

for segment in vessel_segments:
    xs, ys = zip(*segment)
    ax_center.plot(
        xs, ys,
        color="#9D2345",
        linewidth=2,
        solid_capstyle="round",
        zorder=4
    )

# Small decorative dots suggest parasite/pigment sites.
for x, y in [
    (0.43, 0.62), (0.46, 0.67), (0.55, 0.63),
    (0.56, 0.58), (0.48, 0.57), (0.53, 0.66)
]:
    ax_center.add_patch(
        Circle(
            (x, y), 0.009,
            color="#6B315D",
            zorder=5
        )
    )

ax_center.text(
    0.5, 0.39,
    "PLACENTAL MALARIA",
    ha="center", va="center",
    fontsize=14, fontweight="bold",
    color="#17365D"
)

ax_center.text(
    0.5, 0.33,
    "Central exposure / grouping variable",
    ha="center", va="center",
    fontsize=9,
    color="#454545"
)

# Arrows from the central concept toward each data domain.
arrows = [
    ((0.29, 0.72), (0.035, 0.83), domain_colours["Firmicutes"]),
    ((0.29, 0.39), (0.035, 0.17), domain_colours["Cytokines"]),
    ((0.71, 0.72), (0.965, 0.83), domain_colours["Placental infection markers"]),
    ((0.71, 0.39), (0.965, 0.17), domain_colours["Placental hormones"])
]

for start, end, colour in arrows:
    ax_center.add_patch(
        FancyArrowPatch(
            start, end,
            arrowstyle="-|>",
            mutation_scale=18,
            linewidth=2.5,
            color=colour,
            connectionstyle="arc3,rad=0.12",
            zorder=0
        )
    )

fig.suptitle(
    "Placental Malaria: Microbiome, Placental Infection, Cytokines and Hormones",
    fontsize=17,
    fontweight="bold",
    color="#17365D",
    y=0.99
)

fig.text(
    0.5, 0.025,
    "Charts use detected dataset measurements. Standardised biomarker values "
    "are for visual comparison, not raw concentrations or significance tests. "
    "Arrows show conceptual links, not proven causality.",
    ha="center",
    fontsize=9,
    color="#444444",
    wrap=True
)

plt.subplots_adjust(
    top=0.92,
    bottom=0.13
)

# =========================================================
# 6. SAVE AND DISPLAY
# =========================================================

fig.savefig(
    "placental_malaria_biological_relationships.png",
    dpi=300,
    bbox_inches="tight"
)

fig.savefig(
    "placental_malaria_biological_relationships.pdf",
    bbox_inches="tight"
)

plt.show()

print("\nSaved:")
print("placental_malaria_biological_relationships.png")
print("placental_malaria_biological_relationships.pdf")
```

```python
# ============================================================
# PLACENTAL MALARIA: BIOLOGICAL RELATIONSHIP INFOGRAPHIC
# Uses the existing research DataFrame named df
# ============================================================

import re
import textwrap
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from matplotlib.patches import Circle, Ellipse, FancyArrowPatch
from matplotlib.lines import Line2D

try:
    from scipy.stats import t as student_t
except ImportError:
    student_t = None

# ------------------------------------------------------------
# 1. CONFIGURATION AND DATA PREPARATION
# ------------------------------------------------------------

CENTRAL = "Malaria_Group"

if "df" not in globals():
    raise NameError(
        "Your Excel dataset must already be loaded into a DataFrame named df."
    )

if CENTRAL not in df.columns:
    raise KeyError(
        "Malaria_Group was not found. Run print(df.columns.tolist()) "
        "and update CENTRAL to the correct column name."
    )

data = df.loc[df[CENTRAL].notna()].copy()
data[CENTRAL] = data[CENTRAL].astype(str)

groups = sorted(data[CENTRAL].unique())

if len(groups) < 2:
    raise ValueError("At least two malaria groups are required.")

ID_COLUMNS = [
    c for c in data.columns
    if c.lower().endswith("_id")
    or c.lower() in {"id", "participant", "participantid"}
]

all_columns = [
    c for c in data.columns
    if c != CENTRAL and c not in ID_COLUMNS
]


# ------------------------------------------------------------
# 2. IDENTIFY THE BIOLOGICAL VARIABLES CORRECTLY
# ------------------------------------------------------------

def is_fb_ratio(column):
    """Identify an explicit Firmicutes-to-Bacteroidetes ratio."""
    name = column.lower().replace(" ", "_")

    has_both_taxa = (
        "firmicutes" in name and "bacteroidetes" in name
    )

    has_fb_abbreviation = (
        "ratio" in name
        and any(token in name for token in ["f_b", "f-b", "f/b"])
    )

    return has_both_taxa or has_fb_abbreviation


def numeric_values(column):
    return pd.to_numeric(data[column], errors="coerce")


def is_numeric_measure(column):
    values = numeric_values(column)
    return (
        values.notna().sum() >= 3
        and values.nunique(dropna=True) > 1
    )


# Abundances are separated from the ratio variables.
firmicutes_cols = [
    c for c in all_columns
    if "firmicutes" in c.lower()
    and "bacteroidetes" not in c.lower()
    and not is_fb_ratio(c)
]

bacteroidetes_cols = [
    c for c in all_columns
    if "bacteroidetes" in c.lower()
    and "firmicutes" not in c.lower()
    and not is_fb_ratio(c)
]

ratio_cols = [
    c for c in all_columns
    if is_fb_ratio(c)
]


# If the F/B ratio is not already recorded, calculate it from
# matched Firmicutes and Bacteroidetes measurements at the same
# pregnancy timepoint, where both measurements are available.
def pregnancy_stage(column):
    name = column.lower()

    if any(k in name for k in [
        "early", "first_trimester", "first trimester"
    ]):
        return "Early pregnancy"

    if any(k in name for k in [
        "mid", "middle", "second_trimester", "second trimester"
    ]):
        return "Mid-pregnancy"

    if any(k in name for k in [
        "late", "third_trimester", "third trimester"
    ]):
        return "Late pregnancy"

    if "delivery" in name:
        return "At delivery"

    return None


def make_stage_label(column, taxon=None):
    stage = pregnancy_stage(column)

    if stage is not None:
        return stage

    label = column

    if taxon:
        label = re.sub(taxon, "", label, flags=re.IGNORECASE)

    label = label.replace("_", " ")
    label = re.sub(r"\bpercent(age)?\b", "", label, flags=re.IGNORECASE)
    label = re.sub(r"\bratio\b", "", label, flags=re.IGNORECASE)
    label = re.sub(r"[()%]", "", label)
    label = " ".join(label.split())

    return label.title() if label else "Measurement"


if not ratio_cols:
    f_by_stage = {}
    b_by_stage = {}

    for column in firmicutes_cols:
        stage = pregnancy_stage(column)
        if stage is not None and stage not in f_by_stage:
            f_by_stage[stage] = column

    for column in bacteroidetes_cols:
        stage = pregnancy_stage(column)
        if stage is not None and stage not in b_by_stage:
            b_by_stage[stage] = column

    stage_order = [
        "Early pregnancy",
        "Mid-pregnancy",
        "Late pregnancy",
        "At delivery"
    ]

    for stage in stage_order:
        if stage not in f_by_stage or stage not in b_by_stage:
            continue

        f_col = f_by_stage[stage]
        b_col = b_by_stage[stage]

        f_values = pd.to_numeric(data[f_col], errors="coerce")
        b_values = pd.to_numeric(data[b_col], errors="coerce")

        ratio_name = f"Derived_F_B_Ratio_{stage.replace(' ', '_')}"

        # A ratio is undefined when the Bacteroidetes denominator
        # is zero or missing.
        data[ratio_name] = np.where(
            b_values > 0,
            f_values / b_values,
            np.nan
        )

        ratio_cols.append(ratio_name)


# Cytokine measurements.
cytokine_terms = [
    "cytokine", "tnf", "interleukin", "interferon",
    "ifn", "il-", "il_", "il6", "il10",
    "chemokine", "cxcl", "ccl"
]

cytokine_cols = [
    c for c in all_columns
    if any(term in c.lower() for term in cytokine_terms)
]

# Placental hormone measurements.
hormone_terms = [
    "hormone", "progesterone", "estrogen", "oestrogen",
    "estradiol", "estriol", "hcg", "human_chorionic",
    "lactogen", "hpl"
]

hormone_cols = [
    c for c in all_columns
    if any(term in c.lower() for term in hormone_terms)
]

# Direct infection / placental parasite measurements.
placenta_terms = [
    "hemozoin", "parasite_density", "parasitemia",
    "parasitaemia", "placental_infection",
    "placenta_infection", "placental_parasite",
    "placental_malaria", "malaria_density",
    "infection_score", "histology"
]

placenta_cols = [
    c for c in all_columns
    if any(term in c.lower() for term in placenta_terms)
    and c not in hormone_cols
]

# Numerical pregnancy outcomes.
outcome_terms = [
    "birth_weight", "birthweight",
    "gestational_age", "gestation_weeks",
    "birth_length", "head_circumference",
    "placental_weight", "iugr"
]

outcome_cols = [
    c for c in all_columns
    if any(term in c.lower() for term in outcome_terms)
    and c.lower() != "birth_growth_group"
]


print("CENTRAL VARIABLE:", CENTRAL)
print("GROUPS:", groups)

for label, columns in [
    ("Firmicutes abundance", firmicutes_cols),
    ("Bacteroidetes abundance", bacteroidetes_cols),
    ("F/B ratios (existing or derived)", ratio_cols),
    ("Direct placental infection markers", placenta_cols),
    ("Cytokines", cytokine_cols),
    ("Placental hormones", hormone_cols),
    ("Pregnancy outcomes", outcome_cols)
]:
    print(f"\n{label}:")
    print(columns if columns else "No matching columns detected.")


# ------------------------------------------------------------
# 3. PLOT STYLE AND STATISTICAL HELPERS
# ------------------------------------------------------------

plt.rcParams["font.family"] = "DejaVu Sans"
plt.rcParams["axes.facecolor"] = "white"
plt.rcParams["figure.facecolor"] = "white"

group_palette = plt.get_cmap("Set1")

group_colours = {
    group: group_palette(i % 9)
    for i, group in enumerate(groups)
}

domain_colours = {
    "microbiome": "#078D83",
    "placenta": "#D95770",
    "cytokines": "#8964B5",
    "hormones": "#E39A16",
    "outcomes": "#3678A8"
}


def mean_ci(values):
    """Mean and approximate 95% confidence interval using a t distribution."""
    values = pd.Series(values).dropna()
    n = len(values)

    if n == 0:
        return np.nan, np.nan, np.nan

    mean = values.mean()

    if n < 2:
        return mean, mean, mean

    se = values.std(ddof=1) / np.sqrt(n)

    if student_t is not None:
        critical = student_t.ppf(0.975, n - 1)
    else:
        critical = 1.96

    return mean, mean - critical * se, mean + critical * se


def style_panel(ax, title, colour):
    ax.set_title(
        title,
        color=colour,
        fontsize=11,
        fontweight="bold",
        loc="left",
        pad=9
    )

    for spine in ax.spines.values():
        spine.set_color("#DCE3EA")
        spine.set_linewidth(0.8)

    ax.tick_params(labelsize=8, colors="#344054")
    ax.grid(axis="y", alpha=0.20)
    ax.set_axisbelow(True)


def empty_panel(ax, title, colour, message):
    style_panel(ax, title, colour)
    ax.text(
        0.5, 0.48,
        message,
        ha="center",
        va="center",
        fontsize=9,
        color="#606A78",
        wrap=True,
        transform=ax.transAxes
    )
    ax.set_xticks([])
    ax.set_yticks([])


# ------------------------------------------------------------
# 4. FIRMIICUTES AND BACTEROIDETES ABUNDANCE
# ------------------------------------------------------------

def plot_microbiome(ax):
    style_panel(
        ax,
        "GUT MICROBIOME — TAXON ABUNDANCE",
        domain_colours["microbiome"]
    )

    taxa = {
        "Firmicutes": firmicutes_cols,
        "Bacteroidetes": bacteroidetes_cols
    }

    available = {
        name: columns
        for name, columns in taxa.items()
        if columns
    }

    if not available:
        empty_panel(
            ax,
            "GUT MICROBIOME — TAXON ABUNDANCE",
            domain_colours["microbiome"],
            "No Firmicutes or Bacteroidetes abundance columns were detected."
        )
        return

    all_cols = [
        c for columns in available.values()
        for c in columns
    ]

    stages = list(dict.fromkeys(
        make_stage_label(
            c,
            "Firmicutes" if "firmicutes" in c.lower() else "Bacteroidetes"
        )
        for c in all_cols
    ))

    def stage_rank(stage):
        order = {
            "Early pregnancy": 0,
            "Mid-pregnancy": 1,
            "Late pregnancy": 2,
            "At delivery": 3
        }
        return order.get(stage, 10)

    stages = sorted(stages, key=stage_rank)
    stage_x = {stage: i for i, stage in enumerate(stages)}

    styles = {
        "Firmicutes": ("o", "-"),
        "Bacteroidetes": ("s", "--")
    }

    for taxon, columns in available.items():
        marker, linestyle = styles[taxon]

        for group in groups:
            means, lows, highs, xs = [], [], [], []

            group_mask = data[CENTRAL] == group

            for stage in stages:
                matching = [
                    c for c in columns
                    if make_stage_label(
                        c,
                        "Firmicutes" if taxon == "Firmicutes"
                        else "Bacteroidetes"
                    ) == stage
                ]

                if not matching:
                    continue

                block = data.loc[group_mask, matching].apply(
                    pd.to_numeric, errors="coerce"
                )

                # If multiple columns map to one timepoint, use the
                # participant-level mean of those columns.
                participant_values = block.mean(axis=1).dropna()

                if participant_values.empty:
                    continue

                mean, low, high = mean_ci(participant_values)

                xs.append(stage_x[stage])
                means.append(mean)
                lows.append(low)
                highs.append(high)

            if xs:
                yerr = np.array([
                    np.maximum(np.array(means) - np.array(lows), 0),
                    np.maximum(np.array(highs) - np.array(means), 0)
                ])

                ax.errorbar(
                    xs, means,
                    yerr=yerr,
                    color=group_colours[group],
                    marker=marker,
                    linestyle=linestyle,
                    linewidth=1.8,
                    markersize=5,
                    capsize=3,
                    label=f"{group} — {taxon}"
                )

    ax.set_xticks(range(len(stages)))
    ax.set_xticklabels(stages, rotation=15, ha="right", fontsize=8)

    percent_data = all(
        "%" in c or "percent" in c.lower()
        for c in all_cols
    )

    ax.set_ylabel(
        "Measured abundance (%)" if percent_data
        else "Measured abundance (recorded units)",
        fontsize=9
    )

    ax.legend(
        fontsize=7,
        ncol=2,
        frameon=False,
        loc="best"
    )


# ------------------------------------------------------------
# 5. FIRMIICUTES-TO-BACTEROIDETES RATIO — SEPARATE PANEL
# ------------------------------------------------------------

def plot_ratio(ax):
    style_panel(
        ax,
        "FIRMIICUTES : BACTEROIDETES RATIO",
        domain_colours["microbiome"]
    )

    ratio_numeric = [
        c for c in ratio_cols
        if c in data.columns and is_numeric_measure(c)
    ]

    if not ratio_numeric:
        empty_panel(
            ax,
            "FIRMIICUTES : BACTEROIDETES RATIO",
            domain_colours["microbiome"],
            "No usable ratio values were found. A ratio is derived only "
            "when matching taxon measurements are available at the same timepoint."
        )
        return

    labels = [make_stage_label(c) for c in ratio_numeric]
    x = np.arange(len(ratio_numeric))

    for group in groups:
        means, lows, highs = [], [], []

        for col in ratio_numeric:
            values = pd.to_numeric(
                data.loc[data[CENTRAL] == group, col],
                errors="coerce"
            ).dropna()

            mean, low, high = mean_ci(values)
            means.append(mean)
            lows.append(low)
            highs.append(high)

        yerr = np.array([
            np.maximum(np.array(means) - np.array(lows), 0),
            np.maximum(np.array(highs) - np.array(means), 0)
        ])

        ax.errorbar(
            x, means,
            yerr=yerr,
            marker="o",
            linewidth=1.8,
            markersize=5,
            capsize=3,
            color=group_colours[group],
            label=group
        )

    ax.set_xticks(x)
    ax.set_xticklabels(labels, fontsize=7)
    ax.set_ylabel("Mean F/B ratio (unitless)", fontsize=8)
    ax.legend(fontsize=7, frameon=False, ncol=2)


# ------------------------------------------------------------
# 6. STANDARDISED BIOMARKER COMPARISON
# ------------------------------------------------------------

def plot_marker_forest(ax, columns, title, colour):
    """
    Each variable is standardised separately. Plot group mean z-scores
    with 95% confidence intervals, not raw concentrations.
    """

    valid_columns = []

    for col in columns:
        vals = numeric_values(col).dropna()

        if len(vals) >= 3 and vals.std(ddof=0) > 0:
            valid_columns.append(col)

    if not valid_columns:
        empty_panel(
            ax, title, colour,
            "No usable varying numeric measurements were detected "
            "for this biological domain."
        )
        return

    style_panel(ax, title, colour)

    y_positions = np.arange(len(valid_columns))
    group_offsets = np.linspace(
        -0.16, 0.16, len(groups)
    ) if len(groups) > 1 else [0]

    for group_index, group in enumerate(groups):
        for row_index, col in enumerate(valid_columns):

            raw = numeric_values(col)
            overall_mean = raw.mean()
            overall_sd = raw.std(ddof=0)

            z = (raw - overall_mean) / overall_sd
            group_values = z.loc[data[CENTRAL] == group].dropna()

            if group_values.empty:
                continue

            mean, low, high = mean_ci(group_values)
            y = y_positions[row_index] + group_offsets[group_index]

            ax.errorbar(
                mean, y,
                xerr=np.array([[
                    max(0, mean - low)
                ], [
                    max(0, high - mean)
                ]]),
                fmt="o",
                color=group_colours[group],
                ecolor=group_colours[group],
                markersize=4.5,
                capsize=2.5,
                linewidth=1.2
            )

    ax.axvline(0, color="#5A6573", linewidth=1, linestyle="--")

    ax.set_yticks(y_positions)
    ax.set_yticklabels(
        [textwrap.fill(c.replace("_", " "), 25) for c in valid_columns],
        fontsize=7
    )

    ax.set_ylim(len(valid_columns) - 0.5, -0.5)
    ax.set_xlabel(
        "Group mean standardised value (z-score; 95% CI)",
        fontsize=8
    )


# ------------------------------------------------------------
# 7. DIRECT PLACENTAL INFECTION MARKERS
# ------------------------------------------------------------

def plot_placental_markers(ax):
    numeric_markers = [
        c for c in placenta_cols
        if c in data.columns and is_numeric_measure(c)
    ]

    categorical_markers = [
        c for c in placenta_cols
        if c in data.columns and c not in numeric_markers
        and data[c].dropna().nunique() > 1
    ]

    if numeric_markers:
        plot_marker_forest(
            ax,
            numeric_markers,
            "PLACENTAL INFECTION / PARASITE MARKERS",
            domain_colours["placenta"]
        )
        return

    if categorical_markers:
        marker = categorical_markers[0]

        style_panel(
            ax,
            f"PLACENTAL MARKER: {marker.replace('_', ' ')}",
            domain_colours["placenta"]
        )

        tab = pd.crosstab(
            data[marker].astype(str),
            data[CENTRAL]
        )

        tab.plot(
            kind="bar",
            ax=ax,
            color=[group_colours[g] for g in tab.columns],
            width=0.75
        )

        ax.set_xlabel("Recorded marker category", fontsize=8)
        ax.set_ylabel("Participants (n)", fontsize=8)
        ax.tick_params(axis="x", labelrotation=30)
        ax.legend(title="Malaria group", fontsize=7)

        return

    style_panel(
        ax,
        "INFECTED PLACENTA / PARASITE MARKERS",
        domain_colours["placenta"]
    )

    counts = data[CENTRAL].value_counts().reindex(groups, fill_value=0)

    bars = ax.bar(
        groups,
        counts.values,
        color=[group_colours[g] for g in groups],
        edgecolor="white"
    )

    ax.bar_label(bars, padding=3, fontsize=9)
    ax.set_ylabel("Participants (n)", fontsize=9)
    ax.set_xlabel("Recorded malaria group", fontsize=8)

    ax.text(
        0.5, 0.05,
        "Group counts only: no separate numeric or categorical "
        "placental infection marker was detected.",
        transform=ax.transAxes,
        ha="center",
        va="bottom",
        fontsize=7,
        color="#555555",
        wrap=True
    )


# ------------------------------------------------------------
# 8. PREGNANCY OUTCOMES
# ------------------------------------------------------------

def plot_outcomes(ax):
    numeric_outcomes = [
        c for c in outcome_cols
        if c in data.columns and is_numeric_measure(c)
    ]

    if numeric_outcomes:
        plot_marker_forest(
            ax,
            numeric_outcomes,
            "PREGNANCY AND BIRTH OUTCOMES",
            domain_colours["outcomes"]
        )
        return

    growth_col = next(
        (
            c for c in data.columns
            if c.lower().replace(" ", "_") == "birth_growth_group"
        ),
        None
    )

    if growth_col is not None:
        style_panel(
            ax,
            "BIRTH-GROWTH CATEGORIES",
            domain_colours["outcomes"]
        )

        tab = pd.crosstab(
            data[growth_col],
            data[CENTRAL]
        )

        tab.plot(
            kind="bar",
            ax=ax,
            color=[group_colours[g] for g in tab.columns]
        )

        ax.set_ylabel("Participants (n)", fontsize=8)
        ax.set_xlabel("Birth-growth group", fontsize=8)
        ax.tick_params(axis="x", labelrotation=25)
        ax.legend(title="Malaria group", fontsize=7)
        return

    empty_panel(
        ax,
        "PREGNANCY AND BIRTH OUTCOMES",
        domain_colours["outcomes"],
        "No usable numeric pregnancy outcome or birth-growth group "
        "was detected. Check your column names."
    )


# ------------------------------------------------------------
# 9. DRAW THE BIOLOGICAL ILLUSTRATIONS IN THE CENTRE
# ------------------------------------------------------------

def draw_rod(ax, x, y, colour, angle=0):
    ax.add_patch(
        Ellipse(
            (x, y),
            width=0.105,
            height=0.034,
            angle=angle,
            facecolor=colour,
            edgecolor="#344054",
            linewidth=0.6,
            zorder=4
        )
    )


def draw_bacteria_icon(ax, x, y):
    draw_rod(ax, x - 0.035, y + 0.02, "#24A6A0", 30)
    draw_rod(ax, x + 0.035, y, "#5F9EE8", -25)
    draw_rod(ax, x - 0.01, y - 0.03, "#8B70CC", 10)


def draw_molecule_icon(ax, x, y, colour):
    points = [
        (x - 0.04, y + 0.025),
        (x + 0.02, y + 0.035),
        (x + 0.045, y - 0.015),
        (x - 0.015, y - 0.035)
    ]

    for a, b in [
        (0, 1), (1, 2), (2, 3), (3, 0)
    ]:
        ax.plot(
            [points[a][0], points[b][0]],
            [points[a][1], points[b][1]],
            color=colour,
            linewidth=1.7,
            zorder=2
        )

    for px, py in points:
        ax.add_patch(
            Circle(
                (px, py),
                0.014,
                facecolor=colour,
                edgecolor="white",
                linewidth=0.5,
                zorder=3
            )
        )


def draw_infection_icon(ax, x, y):
    ax.add_patch(
        Ellipse(
            (x, y),
            width=0.13,
            height=0.075,
            facecolor="#F5A3AA",
            edgecolor="#BA405C",
            linewidth=1.2
        )
    )

    for px, py in [
        (x - 0.04, y + 0.015),
        (x + 0.01, y + 0.025),
        (x + 0.035, y - 0.015),
        (x - 0.01, y - 0.02)
    ]:
        ax.add_patch(
            Circle(
                (px, py),
                0.009,
                facecolor="#69375E",
                edgecolor="none"
            )
        )


def draw_cytokine_icon(ax, x, y):
    for dx, dy, size in [
        (-0.04, 0.02, 0.012),
        (-0.015, 0.04, 0.009),
        (0.015, 0.025, 0.014),
        (0.04, 0.005, 0.01),
        (-0.02, -0.018, 0.014),
        (0.015, -0.025, 0.009)
    ]:
        ax.add_patch(
            Circle(
                (x + dx, y + dy),
                size,
                facecolor="#9C7BD0",
                edgecolor="white",
                linewidth=0.5
            )
        )


def draw_central_placenta(ax):
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.axis("off")

    # Background halo.
    ax.add_patch(
        Circle(
            (0.5, 0.57),
            0.30,
            facecolor="#FFF0EE",
            edgecolor="#D95770",
            linewidth=2.2,
            zorder=1
        )
    )

    # Simplified placenta shape.
    ax.add_patch(
        Ellipse(
            (0.5, 0.60),
            width=0.46,
            height=0.31,
            angle=-8,
            facecolor="#E98892",
            edgecolor="#A92F4B",
            linewidth=1.8,
            zorder=2
        )
    )

    ax.add_patch(
        Ellipse(
            (0.5, 0.60),
            width=0.35,
            height=0.22,
            angle=-8,
            facecolor="#F7B7AB",
            edgecolor="#C64F65",
            linewidth=1.3,
            zorder=3
        )
    )

    # Illustrative branching vessels.
    vessel_segments = [
        [(0.5, 0.60), (0.46, 0.65), (0.42, 0.67)],
        [(0.49, 0.60), (0.45, 0.58), (0.42, 0.55)],
        [(0.50, 0.60), (0.54, 0.64), (0.58, 0.65)],
        [(0.51, 0.60), (0.55, 0.59), (0.58, 0.56)],
        [(0.50, 0.61), (0.49, 0.57), (0.47, 0.55)],
        [(0.51, 0.60), (0.52, 0.57), (0.55, 0.54)]
    ]

    for segment in vessel_segments:
        xs, ys = zip(*segment)
        ax.plot(
            xs, ys,
            color="#9D2345",
            linewidth=2.0,
            solid_capstyle="round",
            zorder=4
        )

    # Schematic pigment / parasite sites, not actual microscopy.
    for x, y in [
        (0.43, 0.61), (0.46, 0.67), (0.55, 0.63),
        (0.56, 0.58), (0.48, 0.56), (0.53, 0.66)
    ]:
        ax.add_patch(
            Circle(
                (x, y),
                0.009,
                color="#63365A",
                zorder=5
            )
        )

    ax.text(
        0.5, 0.365,
        "PLACENTAL MALARIA",
        fontsize=12,
        fontweight="bold",
        color="#17365D",
        ha="center",
        va="center"
    )

    ax.text(
        0.5, 0.315,
        "Central grouping / exposure variable",
        fontsize=8,
        color="#536173",
        ha="center",
        va="center"
    )

    # Four small biological icons around the organ.
    draw_bacteria_icon(ax, 0.14, 0.88)
    draw_infection_icon(ax, 0.86, 0.88)
    draw_cytokine_icon(ax, 0.14, 0.16)
    draw_molecule_icon(ax, 0.86, 0.16, "#E39A16")

    ax.text(
        0.14, 0.80, "GUT MICROBIOME",
        ha="center", fontsize=7,
        color=domain_colours["microbiome"],
        fontweight="bold"
    )

    ax.text(
        0.86, 0.80, "INFECTION MARKERS",
        ha="center", fontsize=7,
        color=domain_colours["placenta"],
        fontweight="bold"
    )

    ax.text(
        0.14, 0.08, "CYTOKINES",
        ha="center", fontsize=7,
        color=domain_colours["cytokines"],
        fontweight="bold"
    )

    ax.text(
        0.86, 0.08, "HORMONES",
        ha="center", fontsize=7,
        color=domain_colours["hormones"],
        fontweight="bold"
    )


# ------------------------------------------------------------
# 10. ASSEMBLE THE FIGURE
# ------------------------------------------------------------

fig = plt.figure(figsize=(20, 14), dpi=150)

# Axes positions in figure coordinates.
ax_micro = fig.add_axes([0.035, 0.575, 0.29, 0.335])
ax_ratio = fig.add_axes([0.035, 0.425, 0.29, 0.115])
ax_cyt = fig.add_axes([0.035, 0.08, 0.29, 0.29])

ax_placenta = fig.add_axes([0.675, 0.575, 0.29, 0.335])
ax_hormone = fig.add_axes([0.675, 0.08, 0.29, 0.335])

ax_center = fig.add_axes([0.36, 0.33, 0.28, 0.43])
ax_outcomes = fig.add_axes([0.36, 0.07, 0.28, 0.19])

# Data-driven panels.
plot_microbiome(ax_micro)
plot_ratio(ax_ratio)

plot_marker_forest(
    ax_cyt,
    cytokine_cols,
    "CYTOKINES — GROUP COMPARISONS",
    domain_colours["cytokines"]
)

plot_placental_markers(ax_placenta)

plot_marker_forest(
    ax_hormone,
    hormone_cols,
    "PLACENTAL HORMONES — GROUP COMPARISONS",
    domain_colours["hormones"]
)

plot_outcomes(ax_outcomes)

# Central schematic.
draw_central_placenta(ax_center)


# ------------------------------------------------------------
# 11. CONNECT THE PANELS TO THE CENTRAL CONCEPT
# ------------------------------------------------------------

arrow_specs = [
    ((0.365, 0.69), (0.328, 0.75), domain_colours["microbiome"]),
    ((0.635, 0.69), (0.672, 0.75), domain_colours["placenta"]),
    ((0.365, 0.43), (0.328, 0.22), domain_colours["cytokines"]),
    ((0.635, 0.43), (0.672, 0.22), domain_colours["hormones"]),
    ((0.50, 0.325), (0.50, 0.265), domain_colours["outcomes"])
]

for start, end, colour in arrow_specs:
    fig.add_artist(
        FancyArrowPatch(
            start, end,
            transform=fig.transFigure,
            arrowstyle="-|>",
            mutation_scale=14,
            linewidth=1.8,
            color=colour,
            connectionstyle="arc3,rad=0.08",
            zorder=20
        )
    )


# ------------------------------------------------------------
# 12. TITLES, LEGEND AND RESEARCH CAUTION
# ------------------------------------------------------------

fig.suptitle(
    "Placental Malaria: Microbiome, Placental Infection, "
    "Cytokines, Hormones and Pregnancy Outcomes",
    fontsize=16,
    fontweight="bold",
    color="#17365D",
    y=0.992
)

fig.text(
    0.5, 0.967,
    "Exploratory relationships calculated from the available research measurements",
    ha="center",
    fontsize=10,
    color="#596579"
)

group_handles = [
    Line2D(
        [0], [0],
        color=group_colours[group],
        marker="o",
        linewidth=1.8,
        markersize=6,
        label=group
    )
    for group in groups
]

fig.legend(
    handles=group_handles,
    title="Malaria group",
    loc="upper center",
    bbox_to_anchor=(0.5, 0.945),
    ncol=max(1, len(groups)),
    frameon=False,
    fontsize=9
)

fig.text(
    0.5, 0.025,
    "Error bars indicate 95% confidence intervals for group means. "
    "Biomarker forest plots use variable-specific z-scores for comparison. "
    "Illustrations are schematic; arrows indicate conceptual connections, "
    "not demonstrated causal pathways. No p-values are shown.",
    ha="center",
    va="bottom",
    fontsize=8,
    color="#4B5563",
    wrap=True
)

# Export sharp files for reports and presentations.
fig.savefig(
    "placental_malaria_integrated_figure.png",
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)

fig.savefig(
    "placental_malaria_integrated_figure.pdf",
    bbox_inches="tight",
    facecolor="white"
)

plt.show()


# ------------------------------------------------------------
# 13. EXPORT DESCRIPTIVE TABLES
# ------------------------------------------------------------

measurement_columns = list(dict.fromkeys(
    firmicutes_cols
    + bacteroidetes_cols
    + ratio_cols
    + placenta_cols
    + cytokine_cols
    + hormone_cols
    + outcome_cols
))

summary_rows = []

for column in measurement_columns:
    if column not in data.columns:
        continue

    if not is_numeric_measure(column):
        continue

    for group in groups:
        values = pd.to_numeric(
            data.loc[data[CENTRAL] == group, column],
            errors="coerce"
        ).dropna()

        summary_rows.append({
            "Variable": column,
            "Malaria_Group": group,
            "N": len(values),
            "Mean": values.mean() if len(values) else np.nan,
            "SD": values.std(ddof=1) if len(values) > 1 else np.nan,
            "Median": values.median() if len(values) else np.nan
        })

group_summary = pd.DataFrame(summary_rows)

column_map = pd.DataFrame([
    {"Domain": domain, "Column": column}
    for domain, columns in [
        ("Firmicutes abundance", firmicutes_cols),
        ("Bacteroidetes abundance", bacteroidetes_cols),
        ("F/B ratio", ratio_cols),
        ("Placental infection markers", placenta_cols),
        ("Cytokines", cytokine_cols),
        ("Placental hormones", hormone_cols),
        ("Pregnancy outcomes", outcome_cols)
    ]
    for column in columns
])

with pd.ExcelWriter(
    "placental_malaria_integrated_tables.xlsx"
) as writer:

    column_map.to_excel(
        writer,
        sheet_name="Detected_Variables",
        index=False
    )

    group_summary.to_excel(
        writer,
        sheet_name="Group_Summaries",
        index=False
    )

    growth_col = next(
        (
            c for c in data.columns
            if c.lower().replace(" ", "_") == "birth_growth_group"
        ),
        None
    )

    if growth_col is not None:
        growth_counts = pd.crosstab(
            data[CENTRAL],
            data[growth_col],
            margins=True
        )

        growth_counts.to_excel(
            writer,
            sheet_name="Birth_Growth_Counts"
        )

        growth_percentages = (
            pd.crosstab(
                data[CENTRAL],
                data[growth_col],
                normalize="index"
            ) * 100
        ).round(1)

        growth_percentages.to_excel(
            writer,
            sheet_name="Birth_Growth_Percent"
        )

print("\nFinished.")
print("PNG: placental_malaria_integrated_figure.png")
print("PDF: placental_malaria_integrated_figure.pdf")
print("Excel tables: placental_malaria_integrated_tables.xlsx")
```
